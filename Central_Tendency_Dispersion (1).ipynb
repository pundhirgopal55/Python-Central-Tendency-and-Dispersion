```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/diabetes.csv')
```

```python
# Dataset Structure;
#row
print("Rows",df.shape[0])
#Column Name :-
print("\n Column Name :",df.columns)
#datatypes :-
print("\n\nDatatype :-",df.dtypes)
```

```python
df.head(10)
```

```python
#selsect hte Numerical cooumn
numeric_df=df.select_dtypes(include=np.number)
print("Numerical Column:",numeric_df.columns)
```

```python
#MEAN ,MEDIAN , MODE:
# 1.Mean
mean_values = numeric_df.mean()
print("\nMean VAlues :",mean_values)
```

```python
# 2.Meadian
median_values=numeric_df.median()
print("Meadian Values : \n",median_values)
```

```python
# 3.Mode
mode_values=numeric_df.mode().iloc[0]
print("Mode Values : \n",mode_values )
```

```python
# DISPERSION
# Variance , Standard Davation,IQR:RANGE
#1.Sample variance
varience_values=numeric_df.var()
print("Sample Varience : \n",varience_values)
```

```python
#2.Standard Davation
sd_values=numeric_df.std()
print("Standard Davation : \n",sd_values)
```

```python
#3.IQR : Q3-Q1
Q1=numeric_df.quantile(0.25)
Q1
Q3=numeric_df.quantile(0.75)
Q3

```

```python
#3A.Range : Higest Values -Lowest values
higest_values=numeric_df.max()
lowest_values=numeric_df.min()
range_values=higest_values-lowest_values
print("Range : \n",range_values)
```

```python
# Outlier : Upper Boundary : Q3+1.5*IQR
upper_boundary=numeric_df.quantile(0.75)+(1.5*numeric_df.quantile(0.75))
print("Upper Boundary : \n",upper_boundary)
```

```python
# Outlier : Lower Boundary : Q1-1.5*IQR
lower_boundary=numeric_df.quantile(0.25)-(1.5*numeric_df.quantile(0.25))
print("Lower Boundary : \n",lower_boundary)
```
